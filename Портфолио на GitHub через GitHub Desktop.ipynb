```python
# 1. Реальные цены трех квартир (в млн рублей)
real_prices = [10.0, 15.0, 20.0]

# 2. Прогнозы, которые выдала наша необученная модель
predicted_prices = [11.5, 13.0, 24.5]

# 3. Считаем абсолютную ошибку для каждой квартиры вручную
error1 = abs(real_prices[0] - predicted_prices[0])  # abs убирает минус, если он есть
error2 = abs(real_prices[1] - predicted_prices[1])
error3 = abs(real_prices[2] - predicted_prices[2])

# 4. Считаем среднюю ошибку (Математическое среднее)
mae = (error1 + error2 + error3) / 3

# 5. Выводим результат расчетов на экран
print("--- РЕЗУЛЬТАТ МАТЕМАТИЧЕСКОГО АНАЛИЗА ---")
print(f"Ошибка на 1-й квартире: {error1} млн руб.")
print(f"Ошибка на 2-й квартире: {error2} млн руб.")
print(f"Ошибка на 3-й квартире: {error3} млн руб.")
print("---")
print(f"Средняя ошибка прогноза ИИ (MAE): {mae:.2f} млн руб.")
```

```python
import pandas as pd

# 1. Создаем базу данных прямо внутри кода (Имя, Возраст, Класс каюты, Цена билета)
data = {
    "Passenger_Name": ["Braund, Mr. Owen", "Cumings, Mrs. John", "Heikkinen, Miss. Laina", "Futrelle, Mrs. Jacques", "Allen, Mr. William"],
    "Age": [22, 38, 26, 35, 35],
    "Pclass": [3, 1, 3, 1, 3],
    "Fare": [7.25, 71.28, 7.92, 53.10, 8.05]
}

# 2. Превращаем наш словарь в умную таблицу Pandas DataFrame
df = pd.DataFrame(data)

# 3. Выводим таблицу на экран
print("--- НАША ТАБЛИЦА ПАССАЖИРОВ ---")
print(df)

# 4. Пропускаем строку для читаемости
print("\n")

# 5. Просим Pandas автоматически посчитать статистику по ценам и возрасту
print("--- АВТОМАТИЧЕСКАЯ СТАТИСТИКА PANDAS ---")
print(df.describe())
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Задаем площади квартир (в кв. метрах)
X = np.array([[30], [45], [60], [75], [90], [120]])

# 2. Задаем реальную стоимость этих квартир (в млн рублей)
y = np.array([[6.0], [9.0], [12.0], [15.0], [18.0], [24.0]])

# 3. Создаем "умный" алгоритм Линейной Регрессии
model = LinearRegression()

# 4. Обучаем алгоритм на наших данных (ИИ ищет закономерность)
model.fit(X, y)
print("--- МОДЕЛЬ УСПЕШНО ОБУЧЕНА ---")

# 5. Просим ИИ сделать прогноз для абсолютно НОВОЙ квартиры площадью 80 кв.м.
new_apartment = np.array([[80]])
predicted_price = model.predict(new_apartment)

# 6. Выводим предсказание на экран
print("\n--- РЕЗУЛЬТАТ ПРОГНОЗА ИИ ---")
print(f"Для квартиры площадью 80 кв.м. модель предсказала цену: {predicted_price[0][0]:.2f} млн руб.")
```
